```python
import pandas as pd
import numpy as np
```

```python
data_path = "data/winequality-white.csv"
wine_data = pd.read_csv(data_path)
```

```python
wine_data.head()
```

```python
example_entry = wine_data["alcohol"][0]
print(example_entry)
```

```python
# Extract the floating point number in the middle of an "alcohol" entry
def clean_alcohol(entry):
    if (entry[0:2] == "R$") and (entry.count(".") == 1):
        #print(f"entry: {entry}")
        split_entry = entry.split()
        #print(f"split_entry: {split_entry}")
        #print(f"split_entry[1]: {split_entry[1]}")
        numbers = split_entry[1]
        decimal = numbers.split(",") #[0]
        #print(f"decimal[1]: {decimal[1]}")
        return float(decimal[0])
    else:
        return None 
    
print(f"example_entry: {example_entry}")
clean_alcohol(example_entry)
```

```python
# Two different ways to show the same thing
wine_data["cleaned_alcohol"] = wine_data["alcohol"].map(clean_alcohol)
wine_data["cleaned_alcohol"] = [clean_alcohol(x) for x in wine_data["alcohol"]]

#wine_data.head()# , "citric acid", "density", "quality")]# .head(10)
```
