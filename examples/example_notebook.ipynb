```python
def fib(n):
    # 
    if n == 1 or n == 0:
        print(f"Current value of n is {n}")
        print(f"returning 1")
        print(" ")
        return 1
    else:
        print(f"Current value of n is {n}")
        print(f"returing fib({n-1}) + fib({n-2})")
        print(" ")
        return fib(n-1) + fib(n-2)
```
